## 一、常见数据结构和基本语句

### 1. 列表操作
**题目**：创建一个包含10个随机整数的列表（范围1-100），编写代码实现：
- 计算列表所有元素的和
- 找出最大值和最小值
- 将列表逆序排列

```python
import random

# 创建一个包含10个随机整数的列表（范围1-100）
lst = [random.randint(1, 100) for _ in range(10)]
print("原始列表:", lst)

# 计算列表所有元素的和
total = sum(lst)
print("总和:", total)

# 找出最大值和最小值
max_val = max(lst)
min_val = min(lst)
print("最大值:", max_val, "最小值:", min_val)

# 将列表逆序排列
lst.reverse()
print("逆序列表:", lst)
```

### 2. 字典应用
**题目**：创建一个字典，键为学生姓名（至少5个），值为对应的分数（0-100）。编写代码：
- 计算平均分
- 找出最高分和最低分的学生
- 添加一个新学生并更新字典

```python
#your code
scores = {
    "张三": 85,
    "李四": 92,
    "王五": 76,
    "赵六": 88,
    "孙七": 95
}

# 计算平均分
average = sum(scores.values()) / len(scores)
print("平均分:", average)

# 找出最高分和最低分的学生
max_score = max(scores.values())
min_score = min(scores.values())

# 找到分数对应的学生（可能有多个）
max_students = [name for name, score in scores.items() if score == max_score]
min_students = [name for name, score in scores.items() if score == min_score]

print("最高分学生:", max_students, "分数:", max_score)
print("最低分学生:", min_students, "分数:", min_score)

# 添加一个新学生并更新字典
scores["周八"] = 90
print("更新后的字典:", scores)
```

### 4. 条件与循环（素数）
**题目**：打印100以内的所有素数（质数）。

```python
#your code
for num in range(2, 100):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=' ')
print()  # 换行
```

### 5. 回文判断函数
**题目**：编写一个函数 `is_palindrome(s)`，判断字符串 `s` 是否是回文（正反读相同），忽略大小写和非字母数字字符。


```python
#your code
def is_palindrome(s):
    filtered = ''.join(ch.lower() for ch in s if ch.isalnum())
    return filtered == filtered[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("racecar"))  # True
print(is_palindrome("hello"))    # False
```

### 7. 可变参数函数
**题目**：编写函数 `multiply_all(*args)`，接受任意个参数并返回它们的乘积。


```python

#your code

def multiply_all(*args):
    result = 1
    for num in args:
        result *= num
    return result

# 测试
print(multiply_all(2, 3, 4))    # 24
print(multiply_all(5, 6))       # 30
print(multiply_all(10))         # 10
```

## 三、异常处理

### 8. 商和余数
**题目**：定义一个函数`divide(a, b)`，返回整数`a`除整数`b`的商和余数。如果输入非法（除零和类型错误）则捕获 `ValueError` 异常并提示重新输入。


```python
#your code
def divide(a, b):
    try:
        if not (isinstance(a, int) and isinstance(b, int)):
            raise ValueError("输入必须为整数！")
        if b == 0:
            raise ValueError("除数不能为零！")
        return a // b, a % b
    except ValueError as e:
        print(f"发生错误：{e}，请重新输入。")

# 测试
print(divide(10, 3))   # (3, 1)
print(divide(10, 0))   # 发生错误
print(divide(10, 'a')) # 发生错误
```

## 四、文件读写

### 9. 逐行读取统计
**题目**：读取一个`utf-8`文本文件（`data`目录中的文件 `books.txt`），统计文件包含的行数和字符数。


```python
#your code
file_path = 'data/books.txt'
try:
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        line_count = len(lines)
        char_count = sum(len(line) for line in lines)
    print(f"行数：{line_count}")
    print(f"字符数：{char_count}")
except FileNotFoundError:
    print(f"文件 {file_path} 未找到，请确认路径是否正确。")
except Exception as e:
    print(f"读取文件时发生错误：{e}")
```
